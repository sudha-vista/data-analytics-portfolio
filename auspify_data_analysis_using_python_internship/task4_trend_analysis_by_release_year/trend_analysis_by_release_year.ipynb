  ## Objective
  
  The Objective of this project is to analyze how Netflix content production has changed over time. The analysis uses the release year of the content to identify yearly trends and changes in content production.

## Import Required Libraries

First, import the Pandas library for data analysis and Matplotlib for creating the trend chart

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load the Dataset

Load the cleaned Netflix dataset used for the previous tasks.

```python
df = pd.read_csv("cleaned_netflix_dataset.csv")
```

## Explore Release Year

 Take a quick look at the Release_Year column to understand the years available in the dataset.

```python
df["Release_Year"].head()
df["Release_Year"].unique()
```

## Calculate Yearly Content

Releases Count the number of Netflix titles released in each year.

```python
yearly_content = df["Release_Year"].value_counts().sort_index()
yearly_content
```

## Identify Top

Content-Producing Years Find the years with the highest number of Netflix titles.

```python
top_years = yearly_content.sort_values(ascending=False).head(10)
top_years
```

## Create Trend Visualization

 Create a line chart to show how the number of Netflix titles changed over the years.

```python
plt.figure(figsize=(10, 5))

plt.plot(yearly_content.index, yearly_content.values)

plt.title("Netflix Content Production by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

## Compare Growth and Decline

Find the year with the highest number of titles in the dataset.

```python
highest_year = yearly_content.idxmax()
highest_count = yearly_content.max()

print("Year with highest content:", highest_year)
print("Number of titles:", highest_count)
```

## Key Findings

* Netflix content production varies across different release years.

* The number of titles is higher in some years compared to others.

* Recent years show a higher amount of content compared to many earlier years.

* The line chart shows the changes in content production over time.

## Results

The Netflix dataset was successfully analyzed based on release year. The number of titles released each year was calculated, the years with the highest content were identified, and a line chart was created to show the overall trend.
