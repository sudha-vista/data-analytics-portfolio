## Objective
The Objective of this project is to prepare the Netflix dataset for analysis by exploring the data, Identifying missing values, standardizing the column names, converting data types, and exporting the final dataset for future analysis.

## Import Required Libraries

First, import the Pandas library, which is used to load, clean, and analyze the dataset.

```python
import pandas as pd
```

## Load the Dataset

Load the CSV file into a Pandas DataFrame to begin the cleaning process.

```python
df= pd.read_csv("Dataset.csv")
```

## Explore the Dataset

Take a quick look at the dataset to understand its structure, size, columns, and data types.

```python
df.head()

df.shape

df.columns

df.info()
```

## Check Data Quality

Before making any changes, check whether the dataset contains missing values or duplicate records

```python
df.isnull().sum()

df.duplicated().sum()
```

## Standardize Column Names

To improve readablitity and maintain consistency, the column names are formatted by capitalizing the first letter of each word.

```python
df.columns= df.columns.str.title()

df.columns
```

## Convert Data Types

Convert the "date_added" column into datetime format so it can be used easily in future analysis.

```python
df["date_Added"]=pd.to_datetime(df["Date_Added"])
df.head()
```

## Export the Dataset

Save the prepared dataset as a new CSV file for future analysis and visualization.

```python
df.to_csv("Cleaned_netfilx_dataset.csv", index=False)
```
