    The Objective of this project is to analyze the distribution of Movies and TV Shows available on Netflix using Python and Pandas. The content types are compared using counts, percentages, and visualizations.

## import Required Libraries

First, import the Pandas library for data analysis and Matplotlib for creating charts.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load the Dataset

Load the cleaned Netflix dataset into a Pandas DataFrame to begin the analysis.

```python
df=pd.read_csv("Cleaned_netflix_dataset.csv")
```

## Explore the Dataset

Take a quick look at the dataset to understand its structure, size, columns, and content types.

```python
df.head()
df.shape
df.columns
df["Type"].unique()
```

## Analyze Content Types

Find the total number of Movies and TV Shows available in the dataset.

```python
content_counts=df["Type"].value_counts()
content_counts
```

## Calculate Content Proportions

Calculate the percentage of Movies and TV Shows to compare their proportions.

```python
content_percentage=(content_counts/len(df))*100
content_percentage.round(2)
```

## Create Content Distribution Chart

Create a bar chart to compare the number of Movies and TV Shows.

```python
plt.figure(figsize=(8,5))

plt.bar(content_counts.index,content_counts.values)

plt.title("Netflix Content Type Distribution")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

## Create Content Proportion Chart

Create pie chart to show the porportion of Movies and TV Shows.

```python
plt.pie(content_counts, labels=content_counts.index)
plt.show()
```

## Compare Content Types

Compare the number of Movies and TV Shows and find which content type is more common.

```python
most_common_type=content_counts.idxmax()
most_common_count=content_counts.max()

difference=content_counts["Movie"]-content_counts["TV Show"]

print("Most common content type:",most_common_type)
print("Number of titles:",most_common_count)
print("Difference between Movies and TV Shows:",difference)
```

## Key Findings

*   Movies are more common than TV Shows in the dataset.

*   The number of Movies is higher than the number of TV Shows.

*   Movies have the larger share of the total content.

*   TV Shows make up a smaller share of the dataset.

*   The charts clearly show the difference between Movies and TV Shows.
