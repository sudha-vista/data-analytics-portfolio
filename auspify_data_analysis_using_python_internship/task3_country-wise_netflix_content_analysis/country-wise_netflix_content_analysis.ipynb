##Objective

 The Objective of this project is to analyze Netflix content availability across different countries. The country information is cleaned and analyzed to find the countries with the highest number of Netflix titles.



## Import Required Libraries

First, import the Pandas library for data analysis and Matplotlib for creating charts.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load the Dataset

Load the cleaned Netflix dataset used in the previous task.

```python
df=pd.read_csv("cleaned_netflix_dataset.csv")
```

## Explore Country Information

Take a quick look at the Country column to understand the country information available in the dataset.

```python
df["Country"].head()
df["Country"].unique()
```

##Clean Country Information

Some Netflix titles are associated with more than one country. Split the country names and create seperate rows for each country.

```python
country_df=df.copy()

country_df["Country"]=country_df["Country"].astype("string").fillna("Unknown")

country_df["Country"]=country_df["Country"].str.split(",")

country_df=country_df.explode("Country")

country_df["Country"]=country_df["Country"].str.strip()

country_df=country_df[country_df["Country"]!=""]
```

## Calculate Content Count by Country

Calculate the number of Netfilx titles available for each country.

```python
country_counts=country_df["Country"].value_counts()
country_counts.head(10)
```

## Identify Top Content-Producing Countries

Find the top 10 countries with the highest number of Netflix titles.

```python
top_countries=country_counts.head(10)
top_countries
```

## Create Country-wise Chart

Create a bar chart to compare the top 10 countries based on the number of Netflix titles.

```python
plt.figure(figsize=(10,6))

plt.bar(top_countries.index,top_countries.values)

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

##Create Country Ranking

Display the top countries with their ranking and content count.

```python
ranking=top_countries.reset_index()
ranking.columns=["Country","Number of Titles"]

ranking
```

## Generate Business Insights

Identify the country with the highest number of Netflix titles and compare the top countries.

```python
top_country=country_counts.idxmax()
top_country_count=country_counts.max()

print("Top content-producing country:",top_country)
print("Number of titles:",top_country_count)
```

##Key Findings

* The United States is one of the major content-producing countries in the Netflix dataset.

* The top countries contribute a large number of titles to the Netflix content library.

* The country-wise ranking shows differences in content availability across countries.

* The bar chart makes it easier to compare the top content-producing countries.
