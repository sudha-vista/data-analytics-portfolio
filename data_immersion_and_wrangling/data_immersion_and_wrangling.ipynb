## Project Overview
* This project focuses on data cleaning and prepation using a retail sales dataset.
* The main goal is to understand the dataset and make it ready for analysis.

## Objectives
* Understand dataset structure
* Identify missing values and duplicates
* Clean and format the data
* Prepare final dataset for anlaysis

## Dataset Information
* Dataset Name: Retail Sales Dataset
* Format: Excel(.xlsx)
* Contains sales and customer details

## Final Output
* Clean and structured dataset ready for analysis.

**Import Libraries**

```python
import pandas as pd
import numpy as np
```

**Load Dataset**

```python
df = pd.read_csv("retail_sales_dataset.csv")
df.head()
```

**Intial Data Inspection**

```python
df.info()
df.describe()
```

**Fix Column Names**

```python
df.columns=df.columns.str.upper()
df.columns
```

**Check Missing Values**

```python
df.isnull().sum()
```

**Handling Missing Values**

```python
df["SALES_AMOUNT"]=df["SALES_AMOUNT"].fillna(df["SALES_AMOUNT"].mean())
df["LOCATION"]=df["LOCATION"].fillna("unknown")
```

**Outlier Identification and Removal**

```python
df.sort_values(by="SALES_AMOUNT",ascending=False).head()
df=df[df["SALES_AMOUNT"]<1000000]
```

**Check And Remove Duplicate Rows**

```python
df.duplicated().sum()
df=df.drop_duplicates()
```

**Final Validation**

```python
df.info()
df.head()
```

**Save Cleaned File**

```python
df.to_excel("cleaned_retail_sales_dataset.xlsx",index=False)
```
